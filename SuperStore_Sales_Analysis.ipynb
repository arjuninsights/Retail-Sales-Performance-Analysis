```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as nm
import seaborn as sns
```

```python
# upload data set 
df=pd.read_csv('Cleaned_SuperStore_Sales.csv')
```

```python
df.shape
```

```python
# Sales by Category
df.groupby('Category')['Sales'].sum().sort_values(ascending=False).plot(kind='bar', title='Sales By Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
df.groupby('Sub-Category')['Sales'].sum().plot(kind='bar', title='Sales by Sub Category')
plt.xlabel('Sub Category')
plt.ylabel('Sales')
plt.show()
```

```python
# Sales by Region
df.groupby('Region')['Sales'].sum().plot(kind='bar',title='Sales By Region')
```

```python
df.groupby('Segment')['Sales'].sum().plot(kind='bar',title='Sales by Segment')
plt.xlabel('Segment')
plt.ylabel('Sales')
plt.show()
```

```python
df.groupby(['Order_Month_name'])['Sales'].sum().sort_values().plot(kind='bar', title='Sales by Month')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
df.groupby('Order_Day_name')['Sales'].sum().round(2).plot(kind='bar',title='Sales by Day',ylabel='Sales', xlabel='Day')
plt.show()
```

```python
df.groupby('Order_Year')['Sales'].sum().plot(kind='line', legend=True,title='Sales By Year', xlabel='Year', ylabel='Sales')
```

```python
df.groupby('Category')['Profit'].sum().plot(kind='bar')
```

```python
df.groupby('Region')['Profit'].sum().plot(kind='bar')
```

```python
df.groupby('Category').agg(
    {'Profit':'sum',
        'Sales':'sum'}
    ).plot(kind='bar',title='Sale and Profit by Category')
plt.xlabel('Sales')
plt.ylabel('Category')
plt.show()
```

```python
df.groupby('Region').agg(
    {
     'Sales':'sum',
     'Profit':'sum'
    }
).plot(kind='bar', title='Sales and Profit by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```

```python
df.groupby('Category')['Quantity'].count().plot(kind='pie',title='Most Demandable Category')
```
