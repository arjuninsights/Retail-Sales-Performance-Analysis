```python
# |Data Cleaning

# Handing Missing Values
df.isna().sum()
```

```python
# Handing Duplicate Values
df.duplicated().sum()
```

```python
# Handing Date Columns 
print('Before :')
print('Data Type of Order Date : ',df['Order Date'].dtypes)
print('Data Type of Ship Date : ',df['Ship Date'].dtypes)
print('After :')
df['Order Date']=pd.to_datetime(df['Order Date'])
df['Ship Date']=pd.to_datetime(df['Ship Date'])
print('Data Type of Order Date : ',df['Order Date'].dtypes)
print('Data Type of Ship Date : ',df['Ship Date'].dtypes)
```

```python
# Handling Categorical Columns
# Ship Mode,Segment,Country,Region,Category,Sub-Category
print('Values in Ship Mode : ',df['Ship Mode'].unique())
print('Data Type of Ship Mode : ',df['Ship Mode'].unique().dtype)
print('\n')
print('Values in Segment : ',df['Segment'].unique())
print('Data Type of Segment : ',df['Segment'].unique().dtype)
print('\n')
print('Values in Country : ',df['Country'].unique())
print('Data Type of Country : ',df['Country'].unique().dtype)
print('\n')
print('Values in Region : ',df['Region'].unique())
print('Data Type of Region : ',df['Region'].unique().dtype)
print('\n')
print('Values in Category : ',df['Category'].unique())
print('Data Type of Category : ',df['Category'].unique().dtype)
print('\n')
print('Values in Sub-Category : ',df['Sub-Category'].unique())
print('Data Type of Sub-Category : ',df['Sub-Category'].unique().dtype)
```

```python
# Ship Mode,Segment,Country,Region,Category,Sub-Category
df['Ship Mode']=df['Ship Mode'].astype('category')
df['Segment']=df['Segment'].astype('category')
df['Country']=df['Country'].astype('category')
df['Region']=df['Region'].astype('category')
df['Category']=df['Category'].astype('category')
df['Sub-Category']=df['Sub-Category'].astype('category')
```

```python
print('Values in Ship Mode : ',df['Ship Mode'].unique())
print('Data Type of Ship Mode : ',df['Ship Mode'].unique().dtype)
print('\n')
print('Values in Segment : ',df['Segment'].unique())
print('Data Type of Segment : ',df['Segment'].unique().dtype)
print('\n')
print('Values in Country : ',df['Country'].unique())
print('Data Type of Country : ',df['Country'].unique().dtype)
print('\n')
print('Values in Region : ',df['Region'].unique())
print('Data Type of Region : ',df['Region'].unique().dtype)
print('\n')
print('Values in Category : ',df['Category'].unique())
print('Data Type of Category : ',df['Category'].unique().dtype)
print('\n')
print('Values in Sub-Category : ',df['Sub-Category'].unique())
print('Data Type of Sub-Category : ',df['Sub-Category'].unique().dtype)
```

```python
# Now we have checked that there is no duplicate nor missing values in our data 
# We also handle categorical data 
```

```python
# Cheking Data Inconsistency : Whitespace issues or Inconsistent spelling
print(df['Category'].value_counts())
print('\n')
print(df['Region'].value_counts())
print('\n')
print(df['Sub-Category'].value_counts())
print('\n')
print(df['Segment'].value_counts())
print('\n')
print(df['Ship Mode'].value_counts())
print('\n')
print(df['Ship Mode'].value_counts())
```

```python
# We can see all the values in categorical columns are consistance and there is no spelling error or also there is no extra spacing
```

```python
# Creating New Columns
df['Order_Month']=df['Order Date'].dt.month
df['Ship_Month']=df['Ship Date'].dt.month
df['Order_Month_name']=df['Order Date'].dt.month_name()
df['Ship_Month_name']=df['Ship Date'].dt.month_name()
df['Order_Year']=df['Order Date'].dt.year
df['Ship_Year']=df['Ship Date'].dt.year
df['Order_Day']=df['Order Date'].dt.day
df['Order_Day_name']=df['Order Date'].dt.day_name()
df['Ship_Day']=df['Ship Date'].dt.day
df['Ship_Day_name']=df['Ship Date'].dt.day_name()
```

```python
# Feature Engineering
df['Sales_level']=df['Sales'].apply(
    lambda x: 'Very High' if x>10000 else 'High' if x>1000 else 'Low'
)
df['Sales_level'].value_counts()
```

```python
df['Profit_Status']=df['Profit'].apply(
    lambda x: 'High' if x>1000 else 'Low' if x>0 else 'Loss'
)
df['Profit_Status'].value_counts()
```

```python
import numpy as nm
df['Quantity_Type']=nm.where(
    df['Quantity']>=10, 'Bulk Order', 'Regular Order'
)
df['Quantity_Type'].value_counts()
```

```python
df.to_csv('Cleaned_SuperStore_Sales.csv', index=False)
```
